```python
from pathlib import Path
import pandas as pd, numpy as np
from cgaaf.course import Course
# Kurs papkasi / course folder (change to your real course):
COURSE = Path("../demo_course")
if not COURSE.exists():          # synthetic sample, for testing only
    from cgaaf.demo import make_demo
    make_demo(COURSE, n_students=24, log=lambda s: None)
c = Course(COURSE)
print(c.meta)
```

```python
from cgaaf.reports import analytics_report
T, out = analytics_report(c)
print(out)
T.round(3)
```

```python
df = c.dataset()
srl = df.groupby('task_id').agg(plans=('plan_goal_words', lambda s: (s > 0).mean()), reflections=('reflection_words', lambda s: (s > 0).mean()),
                                 mean_plan_min=('plan_minutes', 'mean'), mean_attempts=('attempt', 'max'))
srl.round(3)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(5, 3))
for tid, g in df.groupby('task_id'):
    m = g.groupby('attempt')['score'].mean()
    ax.plot(m.index, m.values, marker='o', label=tid)
ax.set(xlabel='Attempt', ylabel='Mean score', ylim=(0, 100)); ax.legend(frameon=False)
fig.savefig(out / 'fig_score_by_attempt.png', dpi=300, bbox_inches='tight')
```
