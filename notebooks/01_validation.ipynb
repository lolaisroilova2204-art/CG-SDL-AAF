# 01 · Validation of automated scores against expert ratings

**UZ:** Avtomatik ballarning ekspert baholari bilan mosligi (Pearson, Spearman, ICC, Bland–Altman, QWK), ekspertlararo ishonchlilik va har bir metrikaning hissasi.

**EN:** Reproduces Tables/Figures of the *Results – validity* section. Re-run on the real course folder.

```python
from pathlib import Path
import pandas as pd, numpy as np
from cgaaf.course import Course
# Kurs papkasi / course folder (change to your real course):
COURSE = Path("../demo_course")
if not COURSE.exists():          # synthetic sample, for testing only
    from cgaaf.demo import make_demo
    make_demo(COURSE, n_students=24, log=lambda s: None)
c = Course(COURSE)
print(c.meta)
```

```python
from cgaaf.reports import validation_report, results_paragraph
reports = {}
for tid in [None] + [t['id'] for t in c.tasks()]:
    rep, out = validation_report(c, tid, model='stored', lang='en')
    reports[tid or 'all'] = rep
    print(tid or 'all', '→', out)
```

```python
rows = []
for k, rep in reports.items():
    if 'agreement' not in rep: continue
    A = rep['agreement']
    rows.append({'task': k, 'n': A['n'], 'r': A['pearson_r'], 'r 95% CI': f"[{A['pearson_lo']:.3f}, {A['pearson_hi']:.3f}]", 'rho': A['spearman_rho'],
                 'ICC(2,1)': A['icc21'], 'MAE': A['mae'], 'RMSE': A['rmse'], 'BA bias': A['ba_bias'], 'LoA': f"[{A['ba_lo']:.1f}, {A['ba_hi']:.1f}]", 'QWK': A['qwk']})
Table_validity = pd.DataFrame(rows).round(3)
Table_validity
```
